````python
# 함수(keyword를 pages수 만큼 검색한 결과를 데이터프레임에)
def get_items_data(keyword, pages):
    import requests
    from bs4 import BeautifulSoup
    import pandas as pd

    items_list =[] # 크롤링 한 데이터를 담을 list(2차원 리스트)
    for page in range(1,pages+1):
        url = f'https://kin.naver.com/search/list.naver?query={keyword}&page={page}'
        # print(url)
        # 해당 url의 soup객체 -> 글제목, link를 가져와서 items_lis에 append
        response = requests.get(url)
        # print(response.status_code) # 정상 접근 가능한지 여부
        soup = BeautifulSoup(response.text, 'html.parser')
    #     items = soup.select('dt > a')
        dts = soup.find_all('dt')
        # print([item.attrs['href'] for item in items])
    #     for item in items:
    #         items_list.append([item.text, item.attrs.get('href')])
    #         items_list.append({'title':item.text, 'link':item.attrs.get('href')})
        for dt in dts:
            item = dt.find('a')
            items_list.append([item.text, item.attrs.get('href')])

    # print(len(items_list), '개 데이터')
    # print(items_list[:2])
    return pd.DataFrame(items_list, columns=['title', 'link'])
````

````python
# 함수 작성
def get_naver_kin(keyword, display):
    import requests
    import pandas as pd
    import json
    from dotenv import load_dotenv
    import os

    load_dotenv()
    client_id = os.getenv('Client_ID')
    client_secret = os.getenv('Client_Secret')

    url = f'https://openapi.naver.com/v1/search/kin.json?query={keyword}&display={display}'
    headers = {'X-Naver-Client-Id':client_id,
              'X-Naver-Client-Secret':client_secret}
    response = requests.get(url, headers=headers)
    #response.status_code
    # items = json.loads(response.text)['items'] # 방법1과 방법2
    items = response.json()['items'] #방법1에서만 가능한 함수
    items_list = []
    for item in items:
        items_list.append([
            item.get('title').replace('<b>','').replace('</b>',''), # item['title']
            item.get('link'),
            item.get('description').replace('<b>','').replace('</b>','')
        ])
    return pd.DataFrame(items_list, columns=['title', 'link', 'description'])
````

    데이터 : 제목, 이미지링크, 썸네일링크, 높이, 폭

````python
# 방법1
def get_naver_image1(keyword, display):
    import requests
    import pandas as pd
    import json
    from dotenv import load_dotenv
    import os

    load_dotenv()
    client_id = os.getenv('Client_ID')
    client_secret = os.getenv('Client_Secret')
    
    url = f'https://openapi.naver.com/v1/search/image?query={keyword}&display={display}'
    headers = {'X-Naver-Client-Id':client_id,
              'X-Naver-Client-Secret':client_secret}
    response = requests.get(url, headers=headers)
    
    items = response.json()['items']
    
    items_list = []
    for idx, item in enumerate(items):
        items_list.append([
            item.get('title'),
            item.get('link'),
            item.get('thumbnail'),
            item.get('sizeheight'),
            item.get('sizewidth')
        ])
        
        # link와 thumbnail을 저장
        save_image('메인', idx+1, keyword, item.get('link'))
        save_image('썸네일', idx+1, keyword, item.get('thumbnail'))
        # 20%, 40%, ... , 80% 시점 message 출력
        if idx/20 == round(idx/20):
            print(f'~ ~ ~ {idx}% 진행 중입니다 ~ ~ ~')
    print('~ ~ ~ 이미지 파일 저장 완료 ~ ~ ~')
        
    return pd.DataFrame(items_list, columns=['title', 'link', 'thumbnail', 'sizeheight', 'sizewidth'])
````

````python
def save_image(attr, idx, query, url):
    import requests
    '{attr}_{idx}_{query}.{확장자} 이미지 저장'
    file_extension = url.split('.')[-1]
    quote_index = file_extension.find('?')
    
    if quote_index != -1:
        file_extension = file_extension[:quote_index]

    img = requests.get(url).content
    open(f'ch14_image/{attr}_{idx}_{query}.{file_extension}', 'wb').write(img)
````

````python
import requests
from urllib.request import urlopen
from bs4 import BeautifulSoup
import pandas as pd
````

````python
pages = 2
items_list = []
for pageNumber in range(1, (pages+1)):
    url = f'https://www.yes24.com/product/category/bestseller?categoryNumber=001&pageNumber={pageNumber}'
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')  
    items = soup.find_all(class_='itemUnit')
    
    for idx, item in enumerate(items):
        rank = item.select_one('div > em').text
        title = item.select_one('div.info_row.info_name > a.gd_name').text
        authors = item.select('span.authPub.info_auth > a')
        author_list = []
        for author in authors:
            author_list.append(author.text)    
        publisher = item.select_one('span.authPub.info_pub > a').text
        price = item.select_one('div.info_row.info_price > strong').text
        items_list.append({
            '순위': rank,
            '제목': title,
            '저자': author_list,
            '출판': publisher,
            '가격': price
        })
 
bs_df = pd.DataFrame(items_list)
bs_df.to_csv('data/ch14_yes24_top48.csv', index=False, encoding='cp949')
bs_df
````

````python
# 방법1
pages = 2
bestseller_list = []
with open('data/ch14_yes24.txt', 'w', encoding='utf-8') as f:
    pass
for page in range(1, pages+1):
    url = f'https://www.yes24.com/product/category/bestseller?pageNumber={page}'
    response = requests.get(url)
    # print(url)
    # print(response.status_code)
    soup = BeautifulSoup(response.text, 'html.parser')
    ranks_els = soup.select('div.img_upper > em.ico.rank')
    ranks = [int(ranks_el.text) for ranks_el in ranks_els]
    titles_els=soup.select("div.item_info > div.info_row > a.gd_name")
    titles = [titles_el.text for titles_el in titles_els]
    writers_els = soup.select("span.authPub.info_auth")
    # writers_els = soup.find_all('span', class_='info_auth')
    writers = [writers_el.text.strip() for writers_el in writers_els]
    publishers_els = soup.select('div.info_row.info_pubGrp > span.authPub.info_pub')
    publishers = [publishers_el.text for publishers_el in publishers_els]
    prices_els = soup.select("div.info_row > strong.txt_num")
    prices = [prices_el.text for prices_el in prices_els]
    # print(len(ranks), len(titles), len(writers), len(publishers), len(prices))
    with open('data/ch14_yes24.txt', 'a', encoding='utf-8') as f:
        for rank, title, writer, publisher, price in zip(ranks, titles, writers, publishers, prices):
            # print("{},{},{},{},{}".format(rank, title, writer, publisher, price))
            f.write(f'{rank}, {title}, {writer}, {publisher}, {price}\n')
            bestseller_list.append([rank, title, writer, publisher, price])
df = pd.DataFrame(bestseller_list, 
                  columns=['rank', 'title', 'writer', 'publisher', 'price'])
df.to_csv('data/ch14_yes24.csv', index=False)
df
````
