```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3' # TensorFlow 로그 레벨 설정

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB3
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras import regularizers
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ModelCheckpoint, CSVLogger
from tensorflow.keras.preprocessing import image
import numpy as np
import matplotlib.pyplot as plt


# --- 1. 기본 설정 ---
# 데이터 경로 설정
TRAIN_DATA_DIR = 'E:/★★★★★AI★★★★★/음식 ai data/selectStart음식DATA/Computer Vision Lab/train'
VALIDATION_DATA_DIR = 'E:/★★★★★AI★★★★★/음식 ai data/selectStart음식DATA/Computer Vision Lab/valid'
TEST_DATA_DIR = 'E:/★★★★★AI★★★★★/음식 ai data/selectStart음식DATA/Computer Vision Lab/test' # 평가용 데이터셋 경로

# 모델 파라미터
N_CLASSES = 3
IMG_WIDTH, IMG_HEIGHT = 300, 300
BATCH_SIZE = 32
EPOCHS = 30 # 실제 학습 시에는 20~50 정도로 늘리는 것을 추천
LEARNING_RATE = 0.001

# 클래스 라벨 (ImageDataGenerator의 폴더 순서, 보통 알파벳 순)
CLASS_LABELS = ['ramen', 'udon', 'pasta']


# --- 2. 데이터 제너레이터 ---
print("Setting up data generators...")
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

# 검증 및 테스트 데이터에는 데이터 증강을 적용하지 않음 (rescale만 적용)
validation_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    TRAIN_DATA_DIR,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

validation_generator = validation_datagen.flow_from_directory(
    VALIDATION_DATA_DIR,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

# 테스트용 제너레이터 (평가 시 사용)
test_generator = test_datagen.flow_from_directory(
    TEST_DATA_DIR,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False # 평가 시에는 순서가 중요하므로 셔플을 끕니다.
)

print(f"Found {train_generator.samples} training images belonging to {train_generator.num_classes} classes.")
print(f"Found {validation_generator.samples} validation images belonging to {validation_generator.num_classes} classes.")
print(f"Found {test_generator.samples} test images belonging to {test_generator.num_classes} classes.")


# --- 3. 모델 구축 (전이 학습) ---
print("\nBuilding model...")
base_model = EfficientNetB3(weights='imagenet', include_top=False, input_shape=(IMG_WIDTH, IMG_HEIGHT, 3))

# 베이스 모델의 가중치를 동결
for layer in base_model.layers:
    layer.trainable = False

# 새로운 분류기 추가
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(512, activation='relu')(x)
x = Dropout(0.5)(x)
predictions = Dense(N_CLASSES, activation='softmax', kernel_regularizer=regularizers.l2(0.001))(x)

model = Model(inputs=base_model.input, outputs=predictions)

# --- 4. 모델 컴파일 ---
model.compile(optimizer=Adam(learning_rate=LEARNING_RATE),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

model.summary()


# --- 5. 모델 학습 ---
print("\nStarting model training...")
# 최고 성능 모델을 저장할 파일 경로
checkpoint_filepath = 'best_food_classifier_weights.keras' # 파일명 변경 (옵션)
checkpointer = ModelCheckpoint(filepath=checkpoint_filepath,
                               monitor='val_accuracy',
                               verbose=1,
                               save_best_only=True,
                               save_weights_only=True, # 이 줄을 추가합니다.
                               mode='max')

csv_logger = CSVLogger('training_history.log')

history = model.fit(
    train_generator,
    steps_per_epoch=train_generator.samples // BATCH_SIZE,
    validation_data=validation_generator,
    validation_steps=validation_generator.samples // BATCH_SIZE,
    epochs=EPOCHS,
    callbacks=[checkpointer, csv_logger]
)

print("\n--- Training Complete ---")


# ===================================================================
#                       모델 테스트 및 예측 (추가된 부분)
# ===================================================================

# --- 6. 저장된 최고 성능의 모델로 성능 평가 (Evaluate) ---
print("\n--- Evaluating model on the test set ---")

# 모델 구조를 다시 정의하거나, 이미 정의된 'model' 변수를 사용합니다.
# 학습된 가중치를 로드합니다.
model.load_weights(checkpoint_filepath) # <-- 수정된 부분

# 모델 성능 평가
loss, accuracy = model.evaluate(test_generator) # <-- 'best_model' 대신 'model' 사용
print(f"\nTest Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy*100:.2f}%")


# --- 7. 새로운 이미지 하나로 예측 (Predict) ---
print("\n--- Predicting a new image ---")
# 예측할 이미지 경로
IMAGE_TO_PREDICT = 'E:/★★★★★AI★★★★★/음식 ai data/selectStart음식DATA/Computer Vision Lab/test.png'

# 이미지 불러오기 및 전처리
try:
    img = image.load_img(IMAGE_TO_PREDICT, target_size=(IMG_WIDTH, IMG_HEIGHT))
    img_array = image.img_to_array(img)
    img_array /= 255.0
    img_batch = np.expand_dims(img_array, axis=0)

    # 예측 실행
    predictions_single = model.predict(img_batch) # <-- 'best_model' 대신 'model' 사용

    # 결과 해석
    predicted_index = np.argmax(predictions_single[0])
    predicted_label = CLASS_LABELS[predicted_index]
    confidence = np.max(predictions_single[0]) * 100

    # 예측 결과 시각화
    plt.imshow(img)
    plt.title(f"Predicted: {predicted_label} ({confidence:.2f}%)")
    plt.axis('off')
    plt.show()

    print("\n--- Prediction Probabilities ---")
    for i, label in enumerate(CLASS_LABELS):
        print(f"- {label}: {predictions_single[0][i]*100:.2f}%")
except FileNotFoundError:
    print(f"Error: The image file was not found at {IMAGE_TO_PREDICT}")
    print("Please check the file path and try again.")
except Exception as e:
    print(f"An error occurred: {e}")
```
