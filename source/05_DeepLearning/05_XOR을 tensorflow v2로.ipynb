```python
import numpy as np
from tensorflow.keras.models import Sequential # 모델 객체 생성
from tensorflow.keras.layers import Input, Dense # layer층을 쌓을 때 필요
```

```python
# 1. data set
# 2. data 전처리
X_data = np.array([[0,0],
                  [0,1],
                  [1,0],
                  [1,1]])
y_data = np.array([[0],[1],[1],[0]])
```

```python
# 3. 모델 구성
# layer층 X_data 2개 -> 10 -> 20 -> 10 -> 1
model = Sequential()
# model.add(Input(shape(2,))) # 입력층
model.add(Dense(units=10, # 출력
               input_dim=2,
               activation='relu'))
model.add(Dense(units=20, activation='relu'))
model.add(Dense(units=10, activation='relu'))
model.add(Dense(units=1, activation='sigmoid')) 
# 이진분류에서의 출력층 활성화 함수: sigmoid
# 다중분류에서의 출력층 활성화 함수: softmax (출력합을 1로)
model.summary()
```

```python
# 4. 모델 학습과정 설정
model.compile(loss='binary_crossentropy',
             optimizer='adam',
             metrics=['binary_accuracy']) # 2진 분류
```

```python
# 5. 학습
hist = model.fit(X_data, y_data, # 훈력데이터
                epochs=600, #학습횟수
#                 batch_size=4,
                verbose=2) # 학습로그 출력, 0 : 출력안된, 2: 상세 출력
```

```python
# 6. 학습과정 살펴보기
hist.history.keys()
```

```python
import matplotlib.pyplot as plt
fig, loss_ax = plt.subplots(figsize=(10,5))
loss_ax.plot(hist.history['loss'], 'r', label='loss')
loss_ax.set_xlabel('epochs')
loss_ax.set_ylabel('loss')
loss_ax.legend(bbox_to_anchor=(0.99, 0.4))
acc_ax = loss_ax.twinx() # loss_ax와 x축을 공유하는 acc_ax 생성
acc_ax.plot(hist.history['binary_accuracy'], 'b', label='bbinary_accuracy')
acc_ax.set_ylabel('binary_accuracy')
acc_ax.legend(bbox_to_anchor=(0.99, 0.8))
```

```python
# 7. 학습평가하기
score = model.evaluate(X_data, y_data)
print('los :', score[0])
print('binary_accuracy :', score[1])
```

```python
# 8. 모델 사용
model.predict(np.array([[0,1],
                       [1,1]])).round()
```
