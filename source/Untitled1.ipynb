```python
import matplotlib.pyplot as plt
import pandas as pd

# ===== 데이터 준비 =====
df_cnn = pd.DataFrame({
    'Model': ['ResNet', 'EfficientNetB2', 'CNN 앙상블(Eff + Res)'],
    'Accuracy': [0.811, 0.895, 0.906]
})
df_cnn = df_cnn.sort_values(by='Accuracy')

df_ml = pd.DataFrame({
    'Model': ['LightGBM', 'RandomForest', 'XGBoost', 'SVM', 'LR'],
    'CNN_feature Only': [0.802, 0.808, 0.833, 0.865, 0.878],
    'CNN 앙상블 + ML': [0.854, 0.8978, 0.886, 0.8977, 0.894]
})
df_ml = df_ml.sort_values(by='CNN 앙상블 + ML')

# ===== 한글 폰트 설정 =====
plt.rcParams['font.family'] = 'Malgun Gothic'  # Windows 기준
plt.rcParams['axes.unicode_minus'] = False

# ===== 그래프 1: CNN 모델 성능 비교 =====
plt.figure(figsize=(8, 5))
plt.plot(df_cnn['Model'], df_cnn['Accuracy'], marker='o', label='Accuracy', color='blue')
plt.axhline(df_cnn['Accuracy'].max(), color='gray', linestyle='--', linewidth=1, label='최고 정확도')
plt.title('CNN 모델 성능 비교')
plt.ylabel('Accuracy')
plt.ylim(0.75, 0.91)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.savefig('cnn_model_performance.png', dpi=300)
plt.show()

# ===== 그래프 2: CNN Feature 기반 ML 성능 비교 =====
plt.figure(figsize=(10, 6))
plt.plot(df_ml['Model'], df_ml['CNN_feature Only'], marker='o', label='CNN Feature Only', color='orange')
plt.plot(df_ml['Model'], df_ml['CNN 앙상블 + ML'], marker='o', label='CNN 앙상블 + ML', color='green')
plt.title('CNN Feature 기반 ML 성능 비교')
plt.ylabel('Accuracy')
plt.xticks(rotation=15)
plt.ylim(0.75, 0.91)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.savefig('cnn_feature_ml_performance.png', dpi=300)
plt.show()
```
