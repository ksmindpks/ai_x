```python
import os
import pandas as pd

# 1. 폴더 내 모든 엑셀 파일
folder_path = "./레서피"
excel_files = sorted([f for f in os.listdir(folder_path) if f.endswith(".xlsx")])

# 2. 기준 시트명 확보 (첫 파일 기준 최대 8개)
reference_sheets = pd.ExcelFile(os.path.join(folder_path, excel_files[0])).sheet_names[:8]

# 3. 시트별 병합 처리
merged_dataframes = {}

for sheet in reference_sheets:
    frames = []
    for file in excel_files:
        print(file)
        df = pd.read_excel(os.path.join(folder_path, file), sheet_name=sheet, dtype=str)
        df = df.fillna("-")
        df.columns.values[0] = "항목"
        frames.append(df)

    # 하나의 DataFrame으로 합치기
    combined = pd.concat(frames, ignore_index=True)

    # 동일 항목 기준 병합 (값이 '-'가 아닌 것만 병합)
    merged_df = (
        combined.groupby("항목")
        .agg(lambda x: " / ".join(sorted({v for v in x if v != "-"})) if any(v != "-" for v in x) else "-")
        .reset_index()
    )

    merged_dataframes[sheet] = merged_df

# 4. 결과 저장
output_file = os.path.join(folder_path, "병합_레시피_간결버전.xlsx")
with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    for sheet, df in merged_dataframes.items():
        df.to_excel(writer, sheet_name=sheet[:31], index=False)

print(f"[완료] 병합 파일 저장 완료 → {output_file}")
```
